```python
# 📊 Customer Churn Analysis & Customer Intelligence
# By Shruti Vishnoi
# Tools: Python | Pandas | NumPy | SQL | SQLite | Matplotlib | Seaborn

# ============================================================
# 1. IMPORT LIBRARIES AND LOAD DATABASE
# ============================================================

import os
import sqlite3
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

warnings.filterwarnings("ignore")

DB_PATH = "/content/customer_churn.db"

# If the database is not available, upload it in Google Colab.
if not os.path.exists(DB_PATH):
    try:
        from google.colab import files
        print("Database not found. Please upload customer_churn.db.")
        uploaded_files = files.upload()

        uploaded_db = next(
            (filename for filename in uploaded_files if filename.endswith(".db")),
            None
        )

        if uploaded_db is None:
            raise FileNotFoundError(
                "No .db file was uploaded. Please upload customer_churn.db."
            )

        DB_PATH = f"/content/{uploaded_db}"

    except ImportError:
        raise FileNotFoundError(
            f"Database not found at {DB_PATH}. Place the database in the correct path."
        )

conn = sqlite3.connect(DB_PATH)

tables_query = """
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name;
"""

tables = pd.read_sql(tables_query, conn)

print("Available Tables:")
display(tables)

if tables.empty:
    conn.close()
    raise ValueError("No tables were found in the database.")

# Load all tables into a dictionary and create DataFrames.
dataframes = {}

for table_name in tables["name"]:
    safe_table_name = table_name.replace('"', '""')

    df_table = pd.read_sql(
        f'SELECT * FROM "{safe_table_name}"',
        conn
    )

    dataframes[table_name] = df_table
    globals()[f"df_{table_name}"] = df_table

    print(f"Loaded: {table_name}")
    print(f"Shape: {df_table.shape}")
    print()

# Display columns for every table.
for table_name, df_table in dataframes.items():
    print(f"Table Name: {table_name}")
    print("Columns:")
    print(df_table.columns.tolist())
    print()

conn.close()


# ============================================================
# 2. CHECK MISSING VALUES
# ============================================================

for table_name, df_table in dataframes.items():
    print(f"Missing values in {table_name}:")
    display(df_table.isnull().sum())
    print()


# ============================================================
# 3. DATA CLEANING
# ============================================================

# ---------- Customer table ----------
if "db_customer" not in dataframes:
    raise KeyError("The table 'db_customer' was not found in the database.")

df_db_customer = dataframes["db_customer"].copy()

display(df_db_customer.head())
display(df_db_customer.tail())
df_db_customer.info()

# Rename name to customer_name only when the original column exists.
if "name" in df_db_customer.columns:
    df_db_customer.rename(
        columns={"name": "customer_name"},
        inplace=True
    )

# Drop columns only when they exist.
columns_to_drop = [
    column for column in ["interests", "pincode"]
    if column in df_db_customer.columns
]

if columns_to_drop:
    df_db_customer.drop(columns=columns_to_drop, inplace=True)

# Convert DOB to datetime.
if "dob" in df_db_customer.columns:
    df_db_customer["dob"] = pd.to_datetime(
        df_db_customer["dob"],
        errors="coerce"
    )

# Standardize gender values.
if "gender" in df_db_customer.columns:
    df_db_customer["gender"] = df_db_customer["gender"].replace({
        "Men": "Male",
        "Women": "Female"
    })

# Fill missing country values using state-country mapping.
if {"state", "country"}.issubset(df_db_customer.columns):
    state_country_mapping = (
        df_db_customer
        .dropna(subset=["country", "state"])
        .drop_duplicates(subset=["state"])
        .set_index("state")["country"]
        .to_dict()
    )

    df_db_customer["country"] = (
        df_db_customer["country"]
        .fillna(df_db_customer["state"].map(state_country_mapping))
    )

print("Customer table after cleaning:")
display(df_db_customer.head())
print("\nMissing values after customer cleaning:")
display(df_db_customer.isnull().sum())


# ---------- Subscription table ----------
if "db_subscription" not in dataframes:
    raise KeyError("The table 'db_subscription' was not found in the database.")

df_db_subscription = dataframes["db_subscription"].copy()

display(df_db_subscription.head())
df_db_subscription.info()

date_columns = [
    "subscription_start_date",
    "renewal_date",
    "cancellation_date"
]

for column in date_columns:
    if column in df_db_subscription.columns:
        df_db_subscription[column] = pd.to_datetime(
            df_db_subscription[column],
            errors="coerce"
        )

# Create churn_flag.
if "cancellation_date" not in df_db_subscription.columns:
    raise KeyError("The column 'cancellation_date' is required.")

df_db_subscription["churn_flag"] = np.where(
    df_db_subscription["cancellation_date"].notna(),
    1,
    0
)

print("Churn Flag Distribution:")
display(df_db_subscription["churn_flag"].value_counts())


# ---------- Support table ----------
if "db_support" not in dataframes:
    raise KeyError("The table 'db_support' was not found in the database.")

df_db_support = dataframes["db_support"].copy()

display(df_db_support.head())
df_db_support.info()

support_columns_to_drop = [
    column for column in ["col_1", "comment"]
    if column in df_db_support.columns
]

if support_columns_to_drop:
    df_db_support.drop(
        columns=support_columns_to_drop,
        inplace=True
    )

if "complaint_date" in df_db_support.columns:
    df_db_support["complaint_date"] = pd.to_datetime(
        df_db_support["complaint_date"],
        errors="coerce"
    )


# ============================================================
# 4. FEATURE ENGINEERING AND MERGING
# ============================================================

required_id = "customerid"

for table_name, df_table in {
    "customer": df_db_customer,
    "subscription": df_db_subscription,
    "support": df_db_support
}.items():
    if required_id not in df_table.columns:
        raise KeyError(
            f"The column '{required_id}' is missing from the {table_name} table."
        )

print("Unique customer IDs:")
print("Customer:", df_db_customer[required_id].nunique())
print("Subscription:", df_db_subscription[required_id].nunique())
print("Support:", df_db_support[required_id].nunique())

# Create complaint_count before reducing support to one row per customer.
if "complaint_date" in df_db_support.columns:
    df_db_support["complaint_count"] = (
        df_db_support.groupby(required_id)[required_id]
        .transform("count")
    )

    # Keep the latest support record per customer.
    df_db_support = (
        df_db_support
        .sort_values("complaint_date", na_position="first")
        .drop_duplicates(subset=required_id, keep="last")
    )
else:
    df_db_support["complaint_count"] = (
        df_db_support.groupby(required_id)[required_id]
        .transform("count")
    )

    df_db_support = df_db_support.drop_duplicates(
        subset=required_id,
        keep="last"
    )

# Merge subscription and customer tables.
df_merged = df_db_subscription.merge(
    df_db_customer,
    on=required_id,
    how="left",
    suffixes=("", "_customer")
)

# Merge support information.
df = df_merged.merge(
    df_db_support,
    on=required_id,
    how="left",
    suffixes=("", "_support")
)

print("Final merged DataFrame:")
print("Shape:", df.shape)
display(df.head())
print("Columns:")
print(df.columns.tolist())


# ============================================================
# 5. HANDLE MISSING VALUES
# ============================================================

if "escalations" in df.columns:
    df["escalations"] = df["escalations"].fillna("N")

if "csat_score" in df.columns:
    df["csat_score"] = pd.to_numeric(
        df["csat_score"],
        errors="coerce"
    ).fillna(0)

if "complaint_count" in df.columns:
    df["complaint_count"] = pd.to_numeric(
        df["complaint_count"],
        errors="coerce"
    ).fillna(0)

print("Missing values after merging and cleaning:")
display(df.isnull().sum())


# ============================================================
# 6. EXPORT CLEANED DATA
# ============================================================

EXPORT_PATH = "/content/exported_churn_data.csv"
df.to_csv(EXPORT_PATH, index=False)

print(f"Cleaned data exported successfully to: {EXPORT_PATH}")


# ============================================================
# 7. DATA ANALYSIS
# ============================================================

# 1. Churn rate
churn_rate = df["churn_flag"].mean() * 100
print("Churn Rate =", round(churn_rate, 2), "%")

# 2. Retention rate
retention_rate = 100 - churn_rate
print("Retention Rate =", round(retention_rate, 2), "%")

# 3. Churn rate by plan type
if "plan_type" in df.columns:
    churn_by_plan = (
        df.groupby("plan_type")["churn_flag"]
        .mean()
        .mul(100)
        .round(2)
        .reset_index(name="churn_rate_pct")
    )

    print("Churn Rate by Plan Type:")
    display(churn_by_plan)

# 4. Churn by state
if "state" in df.columns:
    churn_by_state = (
        df.groupby("state")["churn_flag"]
        .mean()
        .mul(100)
        .round(2)
        .reset_index(name="churn_rate_pct")
    )

    print("Churn Rate by State:")
    display(churn_by_state)

# 5. ARPU
if "monthly_charges" in df.columns:
    df["monthly_charges"] = pd.to_numeric(
        df["monthly_charges"],
        errors="coerce"
    )

    arpu = df["monthly_charges"].mean()
    print("ARPU =", round(arpu, 2))

# 6. Average customer tenure
if {"subscription_start_date", "cancellation_date"}.issubset(df.columns):
    today = pd.Timestamp.today().normalize()

    end_date = df["cancellation_date"].fillna(today)

    df["tenure_days"] = (
        end_date - df["subscription_start_date"]
    ).dt.days

    avg_tenure = df["tenure_days"].mean()
    print("Average Tenure (Days) =", round(avg_tenure, 2))

# 7. Revenue at risk
if {"monthly_charges", "churn_flag"}.issubset(df.columns):
    revenue_at_risk = df.loc[
        df["churn_flag"] == 1,
        "monthly_charges"
    ].sum()

    print("Revenue at Risk =", round(revenue_at_risk, 2))

# 8. Escalation rate
if "escalations" in df.columns:
    escalation_rate = (
        df["escalations"]
        .astype(str)
        .str.upper()
        .eq("Y")
        .mean()
        * 100
    )

    print("Escalation Rate =", round(escalation_rate, 2), "%")

# 9. Average complaints per user
if {"complaint_count", "customerid"}.issubset(df.columns):
    total_customers = df["customerid"].nunique()

    if total_customers > 0:
        avg_complaints = (
            df["complaint_count"].sum() / total_customers
        )
    else:
        avg_complaints = 0

    print("Average Complaints per User =", round(avg_complaints, 2))

# 10. Correlation between escalation and churn
if {"escalations", "churn_flag"}.issubset(df.columns):
    df["escalations_encoded"] = (
        df["escalations"]
        .astype(str)
        .str.upper()
        .eq("Y")
        .astype(int)
    )

    correlation = df[
        ["escalations_encoded", "churn_flag"]
    ].corr().loc[
        "escalations_encoded",
        "churn_flag"
    ]

    print(
        "Correlation between Escalation and Churn =",
        round(correlation, 2)
    )


# ============================================================
# 8. CHURN RISK FEATURE
# ============================================================

if "churn_score" in df.columns:
    df["churn_score"] = pd.to_numeric(
        df["churn_score"],
        errors="coerce"
    )

    conditions = [
        df["churn_score"] < 50,
        df["churn_score"].between(50, 69, inclusive="both"),
        df["churn_score"] >= 70
    ]

    choices = ["low", "med", "high"]

    df["churn_risk"] = np.select(
        conditions,
        choices,
        default="unknown"
    )
else:
    df["churn_risk"] = "unknown"

df_visual = df.copy()

print("Churn risk distribution:")
display(df_visual["churn_risk"].value_counts())


# ============================================================
# 9. VISUALIZATION USING MATPLOTLIB
# ============================================================

# 9.1 Monthly churn trend
if "cancellation_date" in df_visual.columns:
    df_visual["cancellation_month"] = (
        df_visual["cancellation_date"]
        .dt.to_period("M")
    )

    churn_trend = (
        df_visual[df_visual["churn_flag"] == 1]
        .groupby("cancellation_month")
        .size()
    )

    if not churn_trend.empty:
        plt.figure(figsize=(10, 4))
        plt.plot(
            churn_trend.index.astype(str),
            churn_trend.values,
            marker="o",
            linestyle="--",
            linewidth=2
        )
        plt.title("Monthly Churn Trend")
        plt.xlabel("Month")
        plt.ylabel("Churned Customers")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

# 9.2 Churn rate by plan type
if "plan_type" in df_visual.columns:
    churn_by_plan_plot = (
        df_visual.groupby("plan_type")["churn_flag"]
        .mean()
        .mul(100)
    )

    plt.figure(figsize=(8, 4))
    plt.bar(
        churn_by_plan_plot.index.astype(str),
        churn_by_plan_plot.values
    )
    plt.title("Churn Rate by Plan Type")
    plt.xlabel("Plan Type")
    plt.ylabel("Churn Rate (%)")
    plt.tight_layout()
    plt.show()

# 9.3 Churn rate by state
if "state" in df_visual.columns:
    churn_by_state_plot = (
        df_visual.groupby("state")["churn_flag"]
        .mean()
        .mul(100)
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(12, 5))
    plt.bar(
        churn_by_state_plot.index.astype(str),
        churn_by_state_plot.values
    )
    plt.title("Churn Rate by State")
    plt.xlabel("State")
    plt.ylabel("Churn Rate (%)")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()


# ============================================================
# 10. SEABORN VISUALIZATIONS
# ============================================================

# Prepare a separate encoded DataFrame.
encoding_columns = [
    column for column in [
        "plan_type",
        "contract_type",
        "churn_score",
        "churn_flag",
        "churn_risk",
        "escalations_encoded"
    ]
    if column in df_visual.columns
]

df_encoded = df_visual[encoding_columns].copy()

categorical_columns = [
    column for column in [
        "plan_type",
        "contract_type",
        "churn_risk"
    ]
    if column in df_encoded.columns
]

# Use explicit ordered categories where possible.
order_mappings = {
    "plan_type": ["Basic", "Standard", "Premium"],
    "contract_type": ["Monthly", "Annual"],
    "churn_risk": ["low", "med", "high", "unknown"]
}

for column in categorical_columns:
    if column in order_mappings:
        categories = order_mappings[column]
        df_encoded[column] = pd.Categorical(
            df_encoded[column],
            categories=categories,
            ordered=True
        ).codes
    else:
        df_encoded[column] = (
            df_encoded[column]
            .astype("category")
            .cat.codes
        )

if not df_encoded.empty:
    plt.figure(figsize=(10, 6))
    sns.heatmap(
        df_encoded.corr(numeric_only=True),
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )
    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()

# Pairplot only when there are enough numeric columns.
if len(df_encoded.columns) >= 2:
    sns.pairplot(df_encoded)
    plt.show()

# Catplot only when all required columns exist.
required_catplot_columns = {
    "plan_type",
    "monthly_charges",
    "gender",
    "churn_risk"
}

if required_catplot_columns.issubset(df_visual.columns):
    sns.catplot(
        data=df_visual,
        x="plan_type",
        y="monthly_charges",
        hue="gender",
        col="churn_risk",
        kind="box"
    )
    plt.show()


# ============================================================
# 11. PIVOT TABLES
# ============================================================

if {"plan_type", "churn_flag"}.issubset(df_visual.columns):
    pivot_churn = pd.pivot_table(
        df_visual,
        index="plan_type",
        values="churn_flag",
        aggfunc="mean"
    )

    print("Churn rate pivot table:")
    display(pivot_churn)

if {
    "plan_type",
    "monthly_charges",
    "customerid",
    "churn_flag"
}.issubset(df_visual.columns):

    pivot_summary = pd.pivot_table(
        df_visual,
        index="plan_type",
        values=[
            "monthly_charges",
            "customerid",
            "churn_flag"
        ],
        aggfunc={
            "monthly_charges": "sum",
            "customerid": "nunique",
            "churn_flag": "mean"
        }
    )

    print("Plan summary pivot table:")
    display(pivot_summary)


# ============================================================
# 12. SQL WITH PYTHON - PRACTICE EXAMPLE
# ============================================================

TEST_DB_PATH = "/content/test_database.sqlite"

conn = sqlite3.connect(TEST_DB_PATH)

conn.execute("DROP TABLE IF EXISTS users")

conn.execute("""
CREATE TABLE users (
    first_name TEXT,
    country TEXT,
    budget INTEGER
)
""")

conn.commit()
print("Table created successfully!")

insert_query = """
INSERT INTO users (first_name, country, budget)
VALUES
    ('Madhav', 'India', 5000),
    ('Rishabh', 'Germany', 2500),
    ('Vishakha', 'India', 3500)
"""

conn.execute(insert_query)
conn.commit()
print("Data inserted successfully!")

df_results = pd.read_sql(
    "SELECT * FROM users",
    conn
)

display(df_results)

df_agg = pd.read_sql(
    """
    SELECT country, SUM(budget) AS total_budget
    FROM users
    GROUP BY country
    """,
    conn
)

print("Budget aggregation:")
display(df_agg)

budget_by_country = (
    df_results
    .groupby("country", as_index=False)["budget"]
    .sum()
)

plt.figure(figsize=(8, 5))
sns.barplot(
    x="country",
    y="budget",
    data=budget_by_country
)
plt.title("Total Budget by Country")
plt.xlabel("Country")
plt.ylabel("Total Budget")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

conn.close()
print("SQLite connection closed.")


# ============================================================
# PROJECT COMPLETED
# ============================================================

print("Customer Churn Analysis & Customer Intelligence completed successfully!")
```
